```python
# Use the same HTTP, environment, JSON, and table tools as in class.
import json
import os
from pathlib import Path

import pandas as pd
import requests
from dotenv import load_dotenv
from io import StringIO

HEADERS = {"User-Agent": "ESMT MAAI data wrangling project"}
```

## Source 1 - API

```python
OECD_URL = "https://sdmx.oecd.org/public/rest/data/OECD.SDD.STES,DSD_STES@DF_FINMARK/DEU.A.SHARE......"

r = requests.get(OECD_URL, params={"startPeriod": 2014, "endPeriod": 2024, "format": "csvfile"}, headers=HEADERS, timeout=30)
r.raise_for_status()
print(r.status_code, r.headers["content-type"])

oecd_sample = pd.read_csv(StringIO(r.text))
oecd_sample
```

## Source 2

```python
WB_URL = "https://api.worldbank.org/v2/country/all/indicator/FP.CPI.TOTL.ZG"

r = requests.get(WB_URL, params={"format": "json", "date": "2014:2024", "per_page": 50}, headers=HEADERS, timeout=30)
r.raise_for_status()

meta = r.json()[0]
records = r.json()[1]
meta
```

```python
wb_sample = pd.json_normalize(records)
wb_sample.head(10)
```

## Source 3

```python
WIKI_URL = "https://en.wikipedia.org/wiki/ISO_3166-1"

r = requests.get(WIKI_URL, headers=HEADERS, timeout=30)
r.raise_for_status()
html = r.text

tables = pd.read_html(StringIO(html), match="English short name", keep_default_na=False)
print(len(tables), "table(s) found")
iso_sample = tables[0]
iso_sample.head(10)
```
