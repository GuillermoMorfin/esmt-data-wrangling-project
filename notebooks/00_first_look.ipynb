## Source 1 - API

```python
# Combine the API server address with the search endpoint.
BASE_URL = "https://api.kleinanzeigen-agent.de"
ENDPOINT = "/api/v2/kleinanzeigen/search"
URL = BASE_URL + ENDPOINT

# Start with one generation so the asking prices are easier to compare.
SEARCH_TERM = "iPhone" #it is a common product
params = {
    "q": SEARCH_TERM,
    "category_id": "173",
    "page": 0,
    "size": 100,
}

# Preview the URL locally before sending the credit-consuming request.
# The credits are expensive :((
prepared = requests.Request("GET", URL, params=params).prepare()
print(prepared.url)

# we have the key in the .env file for the API
load_dotenv("../.env")
key = os.getenv("KLAZ_API_KEY", "").strip()

# Stop early if the key or .env path needs to be corrected.
#if not key:
#    raise ValueError("KLAZ_API_KEY is missing. Check the .env path and variable name.")

# Use the authentication header required by this API.
headers = {"klaz_key": key}

# Send the search and keep the response for the next cell.
r = requests.get(URL, headers=headers, params=params, timeout=20)

# Inspect the response before trying to read its JSON.
print("HTTP status:", r.status_code)
print("Content-Type:", r.headers.get("Content-Type"))
if not r.ok:
    print(r.text[:500])
r.raise_for_status()

# Parse the response and inspect the outer dictionary.
response_data = r.json()
print("Response keys:", list(response_data))

# Separate the listing records from the pagination metadata.
data = response_data["data"]
ads = data["ads"]
meta = data["meta"]

df_ads = pd.json_normalize(ads)
df_ads.head()
```
